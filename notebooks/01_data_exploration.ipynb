```python
# Create a copy of the original dataset
df_clean = df.copy()

# 1. Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# 2. Remove cancelled transactions
df_clean = df_clean[
    ~df_clean['InvoiceNo'].astype(str).str.startswith('C')
]

# 3. Keep only valid quantities
df_clean = df_clean[df_clean['Quantity'] > 0]

# 4. Keep only valid prices
df_clean = df_clean[df_clean['UnitPrice'] > 0]

# 5. Remove rows without CustomerID
df_clean = df_clean.dropna(subset=['CustomerID'])

# 6. Remove rows without product description
df_clean = df_clean.dropna(subset=['Description'])

# Reset index
df_clean = df_clean.reset_index(drop=True)

print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
print("Rows removed:", len(df) - len(df_clean))
```

```python
print("Unique Customers:", df_clean['CustomerID'].nunique())
print("Unique Products:", df_clean['StockCode'].nunique())

print("\nCountries:", df_clean['Country'].nunique())

print("\nDate Range:")
print("From:", df_clean['InvoiceDate'].min())
print("To:", df_clean['InvoiceDate'].max())
```

```python
# Create customer-product interaction matrix
interaction_matrix = df_clean.pivot_table(
    index='CustomerID',
    columns='StockCode',
    values='Quantity',
    aggfunc='sum',
    fill_value=0
)

print("Interaction Matrix Shape:", interaction_matrix.shape)

interaction_matrix.head()
```

```python
# Convert purchase quantities into binary interactions
interaction_binary = (interaction_matrix > 0).astype(int)

print("Binary Interaction Matrix Shape:", interaction_binary.shape)

print("\nTotal Purchases/Interactions:",
      interaction_binary.values.sum())
```

```python
# Create train and test interactions
train_matrix = interaction_binary.copy()
test_matrix = interaction_binary.copy()

# For each customer, keep their last purchased product for testing
for customer in df_clean['CustomerID'].unique():
    
    customer_data = df_clean[df_clean['CustomerID'] == customer]
    
    # Last purchased product
    last_product = customer_data.sort_values('InvoiceDate').iloc[-1]['StockCode']
    
    # Remove it from training data
    if last_product in train_matrix.columns:
        train_matrix.loc[customer, last_product] = 0

print("Training Matrix Shape:", train_matrix.shape)
print("Test Matrix Shape:", test_matrix.shape)
```

```python
from sklearn.decomposition import TruncatedSVD

# Convert all product column names to strings
train_matrix.columns = train_matrix.columns.astype(str)

# Number of latent factors
n_components = 50

# Create SVD model
svd = TruncatedSVD(
    n_components=n_components,
    random_state=42
)

# Train the model
customer_factors = svd.fit_transform(train_matrix)

# Product factors
product_factors = svd.components_

print("Customer Factors Shape:", customer_factors.shape)
print("Product Factors Shape:", product_factors.shape)

print("\nExplained Variance Ratio:",
      svd.explained_variance_ratio_.sum())
```

```python
# Generate predicted interaction scores
predicted_scores = customer_factors @ product_factors

print("Predicted Scores Shape:", predicted_scores.shape)
```

```python
# Function to generate recommendations
def get_recommendations(customer_id, top_n=10):
    
    # Get customer position
    customer_index = interaction_binary.index.get_loc(customer_id)
    
    # Get predicted scores for this customer
    scores = predicted_scores[customer_index]
    
    # Get products already purchased
    purchased_products = interaction_binary.loc[customer_id]
    
    # Copy scores so we can modify them
    recommendation_scores = pd.Series(
        scores,
        index=train_matrix.columns
    )
    
    # Remove already purchased products
    recommendation_scores[purchased_products.values == 1] = -1
    
    # Get top recommendations
    top_products = recommendation_scores.sort_values(
        ascending=False
    ).head(top_n)
    
    return top_products
    # Test recommendation for first customer
customer_id = interaction_binary.index[0]

recommendations = get_recommendations(customer_id, top_n=10)

print("Customer ID:", customer_id)
print("\nRecommended Products:")
print(recommendations)
```

```python
import numpy as np

def precision_at_k(actual, recommended, k=10):
    recommended = recommended[:k]
    if len(recommended) == 0:
        return 0.0
    
    hits = len(set(actual) & set(recommended))
    return hits / k


def recall_at_k(actual, recommended, k=10):
    recommended = recommended[:k]
    if len(actual) == 0:
        return 0.0
    
    hits = len(set(actual) & set(recommended))
    return hits / len(actual)


def ndcg_at_k(actual, recommended, k=10):
    recommended = recommended[:k]
    
    relevance = [
        1 if item in actual else 0
        for item in recommended
    ]
    
    dcg = sum(
        rel / np.log2(i + 2)
        for i, rel in enumerate(relevance)
    )
    
    ideal_relevance = sorted(relevance, reverse=True)
    
    idcg = sum(
        rel / np.log2(i + 2)
        for i, rel in enumerate(ideal_relevance)
    )
    
    if idcg == 0:
        return 0.0
    
    return dcg / idcg
    
```

```python
# Evaluate the recommendation model

k = 10

precision_scores = []
recall_scores = []
ndcg_scores = []

for customer_id in interaction_binary.index:

    # Products actually purchased by customer
    actual_products = set(
        test_matrix.loc[customer_id][
            test_matrix.loc[customer_id] == 1
        ].index
    )

    # We only evaluate customers with at least one purchase
    if len(actual_products) == 0:
        continue

    # Get recommendations
    recs = get_recommendations(
        customer_id,
        top_n=k
    )

    recommended_products = list(recs.index)

    # Calculate metrics
    precision_scores.append(
        precision_at_k(
            actual_products,
            recommended_products,
            k
        )
    )

    recall_scores.append(
        recall_at_k(
            actual_products,
            recommended_products,
            k
        )
    )

    ndcg_scores.append(
        ndcg_at_k(
            actual_products,
            recommended_products,
            k
        )
    )

# Average metrics
avg_precision = np.mean(precision_scores)
avg_recall = np.mean(recall_scores)
avg_ndcg = np.mean(ndcg_scores)

print("Model Evaluation Results")
print("------------------------")
print(f"Precision@10: {avg_precision:.4f}")
print(f"Recall@10:    {avg_recall:.4f}")
print(f"NDCG@10:      {avg_ndcg:.4f}")
```

```python
# Evaluate Popularity Baseline

baseline_precision = []
baseline_recall = []
baseline_ndcg = []

for customer_id, actual_item in test_items.items():

    # Remove products already purchased by the customer
    purchased = set(
        train_matrix.loc[customer_id][
            train_matrix.loc[customer_id] == 1
        ].index.astype(str)
    )

    # Select top popular products not already purchased
    baseline_recommendations = [
        product
        for product in popular_products
        if product not in purchased
    ][:10]

    baseline_precision.append(
        precision_at_k(
            [actual_item],
            baseline_recommendations,
            k=10
        )
    )

    baseline_recall.append(
        recall_at_k(
            [actual_item],
            baseline_recommendations,
            k=10
        )
    )

    baseline_ndcg.append(
        ndcg_at_k(
            [actual_item],
            baseline_recommendations,
            k=10
        )
    )


print("Popularity Baseline Results")
print("---------------------------")
print(f"Precision@10: {np.mean(baseline_precision):.4f}")
print(f"Recall@10:    {np.mean(baseline_recall):.4f}")
print(f"NDCG@10:      {np.mean(baseline_ndcg):.4f}")
```
